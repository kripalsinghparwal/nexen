```python
import requests
import os
import time
import random
import json
import pandas as pd
from bs4 import BeautifulSoup

# Load data
df = pd.read_excel(r"D:\Nexensus_Projects\Incorporation\input_folder\Shivam Sir incorporation.xlsx")[:5]

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)",
    "Accept-Language": "en-US,en;q=0.9",
    "Referer": "https://www.tofler.in/"
}

output_file = "output.txt"

with open(output_file, "w", encoding="utf-8") as f:
    for row in df.itertuples(index=False):
        cin = row.cin
        input_name = row.name  
        url = f"https://www.tofler.in/company/{cin}"

        try:
            response = requests.get(url, headers=headers, timeout=60)
            soup = BeautifulSoup(response.text, 'html.parser')

            tag = soup.find(class_="accordion_content p-20 rounded-md")
            incorporation = tag.text.strip() if tag else None

            # Create JSON line
            record = {
                "name": input_name,
                "cin": cin,
                "incorporation": incorporation
            }

            # Write as JSON line
            f.write(json.dumps(record, ensure_ascii=False) + "\n")

            print(f"✅ Done: {cin}")

        except Exception as e:
            print(f"❌ Error for {cin}: {e}")

        time.sleep(random.uniform(2, 4))  # polite delay
```

```python
####################### Code to combine all csv files downloaded for all states ######################
import pandas as pd
import os

# 🔹 Input and Output paths
input_folder = r"D:\Nexensus_Projects\Incorporation\output_folder\downloaded_company_csv"   # change this
output_file = r"merged.csv"

# 🔹 Get all CSV files
csv_files = [f for f in os.listdir(input_folder) if f.endswith(".csv")]

# 🔹 List to store dataframes
df_list = []

for file in csv_files:
    file_path = os.path.join(input_folder, file)
    try:
        df = pd.read_csv(file_path)
        df["source_file"] = file   # optional: track source
        df_list.append(df)
        print(f"✅ Loaded: {file}")
    except Exception as e:
        print(f"❌ Skipped: {file} | Error: {e}")

# 🔹 Merge all dataframes
if df_list:
    merged_df = pd.concat(df_list, ignore_index=True)

    # 🔹 Save to CSV
    merged_df = pd.concat(df_list, ignore_index=True, sort=False)
    print(f"\n🎉 Merged file saved at: {output_file}")
else:
    print("⚠️ No valid CSV files found.")

merged_df.drop_duplicates().to_csv("master.csv")
```
